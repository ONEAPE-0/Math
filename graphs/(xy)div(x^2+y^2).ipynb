$$
\frac{xy}{x^2 + y^2}
$$

```python
# Jupyter script to graph f(x, y) = xy / (x^2 + y^2) interactively with plotly

import numpy as np
import plotly.graph_objects as go

# Define the function
def f(x, y):
    return np.divide(x * y, x**2 + y**2, where=(x**2 + y**2) != 0)  # Avoid division by zero

# Create a grid of x and y values
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Create the interactive plot using Plotly
fig = go.Figure(data=[go.Surface(z=Z, x=X, y=Y, colorscale='Viridis')])

# Update layout for better visualization
fig.update_layout(
    #title=r"$f(x, y) = \frac{xy}{x^2 + y^2}$",
    scene=dict(
        xaxis_title="x",
        yaxis_title="y",
        zaxis_title="f(x, y)"
    )
)

# Display the plot
fig.show()
```
